# MTG Input Generator (Colab)

This notebook runs `src/build_metagame_input.py` and exports ready-to-use files for meta analysis.

```python
# Setup
import os
import subprocess

REPO_URL = "https://github.com/Warlord1986pl/MTG-Metagame-Analyzer.git"
REPO_DIR = "/content/MTG-Metagame-Analyzer"

if not os.path.exists(REPO_DIR):
    subprocess.run(["git", "clone", REPO_URL, REPO_DIR], check=True)

os.chdir(REPO_DIR)
subprocess.run(["python", "-m", "pip", "install", "-q", "-r", "requirements.txt"], check=True)
print("Setup complete:", os.getcwd())
```

```python
# Parameters
format_name = "Modern"
history_points = 4
metagame_window_days = 14
anchor_sunday = "2026-03-15"  # YYYY-MM-DD, or set to None for latest Sunday

my_deck = "Domain Zoo"
my_window_days = 90
my_fallback_window_days = 180
rogue_threshold = 0.5  # Meta % cutoff for Rogue bucket
```

```python
# Run generator
cmd = [
    "python", "src/build_metagame_input.py",
    "--format", format_name,
    "--history-points", str(history_points),
    "--metagame-window-days", str(metagame_window_days),
    "--my-deck", my_deck,
    "--my-window-days", str(my_window_days),
    "--my-fallback-window-days", str(my_fallback_window_days),
    "--rogue-threshold", str(rogue_threshold),
]

if anchor_sunday:
    cmd.extend(["--anchor-sunday", anchor_sunday])

print("Running:", " ".join(cmd))
subprocess.run(cmd, check=True)

# Inspect outputs
import pathlib

out_dir = pathlib.Path("outputs")
for p in sorted(out_dir.rglob("*")):
    if p.is_file() and ("history" in str(p) or p.parent == out_dir):
        print(p)
```

```python
# Download outputs as ZIP
from google.colab import files
import shutil

zip_path = "/content/metagame_outputs.zip"
if os.path.exists(zip_path):
    os.remove(zip_path)

shutil.make_archive("/content/metagame_outputs", "zip", root_dir="outputs")
files.download(zip_path)
```
